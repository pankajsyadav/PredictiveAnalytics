### Introduction of Topic/Problem
Customer churn is a critical business challenge for telecommunications providers, where customer acquisition costs typically exceed retention costs. This project applies predictive analytics to identify customers at high risk of churn before cancellation occurs. The objective is to support proactive retention decisions by combining supervised classification and customer segmentation techniques. The work is aligned with CRISP-DM and emphasizes business-relevant evaluation metrics for imbalanced classification.

## Overview of Data Used
The project uses the Telco Customer Churn dataset from Hugging Face. The target variable is `Churn` (0 = retained, 1 = churned). Predictors include contract, tenure, service usage, payment methods, and billing variables. Training and test data are loaded separately to prevent leakage. Missing values are handled using business-valid assumptions (for example, `No Offer`, `No Internet Service`) where appropriate.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import GridSearchCV
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, classification_report, roc_curve, auc
)
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings("ignore")
```

```python
# Load train and test sets from Hugging Face
base = "hf://datasets/aai510-group1/telco-customer-churn/"
churn_train = pd.read_csv(base + "train.csv")
churn_test = pd.read_csv(base + "test.csv")

print("Train shape:", churn_train.shape)
print("Test shape :", churn_test.shape)
```

```python
print("Target distribution (train):")
print(churn_train["Churn"].value_counts(dropna=False))
print("\nTarget proportion (train):")
print(churn_train["Churn"].value_counts(normalize=True).round(4))

print("\nTop missing columns (train):")
print(churn_train.isnull().sum().sort_values(ascending=False).head(15))
```

## Methods of Analysis: Data Preparation
The analysis begins by removing leakage-prone fields that are only known after churn has already occurred. Missing values in `Offer` and `Internet Type` are filled with business-meaningful categories. Features are then separated from the target and one-hot encoded for model readiness. Train and test feature matrices are aligned to ensure identical columns and correct model scoring.

```python
# Remove post-outcome leakage columns
leakage_cols = ["Churn Reason", "Churn Category", "Churn Score", "Customer Status", "Satisfaction Score"]

for col in leakage_cols:
    if col in churn_train.columns:
        churn_train.drop(columns=col, inplace=True)
    if col in churn_test.columns:
        churn_test.drop(columns=col, inplace=True)

# Fill business-state missing values
if "Offer" in churn_train.columns:
    churn_train["Offer"] = churn_train["Offer"].fillna("No Offer")
    churn_test["Offer"] = churn_test["Offer"].fillna("No Offer")

if "Internet Type" in churn_train.columns:
    churn_train["Internet Type"] = churn_train["Internet Type"].fillna("No Internet Service")
    churn_test["Internet Type"] = churn_test["Internet Type"].fillna("No Internet Service")

print("Remaining missing values (top 10):")
print(churn_train.isnull().sum().sort_values(ascending=False).head(10))
```

```python
# Feature/target split
y_train = churn_train["Churn"].astype(int)
y_test = churn_test["Churn"].astype(int)

X_train = churn_train.drop(columns=["Churn"])
X_test = churn_test.drop(columns=["Churn"])

# One-hot encoding
X_train = pd.get_dummies(X_train, drop_first=True)
X_test = pd.get_dummies(X_test, drop_first=True)

# Align columns
X_train, X_test = X_train.align(X_test, join="left", axis=1, fill_value=0)

print("Encoded X_train shape:", X_train.shape)
print("Encoded X_test shape :", X_test.shape)
```

EDA is used to confirm class imbalance and identify key churn patterns before model training. Contract type, internet service type, tenure, and monthly charges are explored due to their expected business impact. Numeric correlation structure is also reviewed to understand feature overlap and support model interpretation. This stage informs both feature strategy and metric selection.

## Results Interpretation: Three Key EDA Views

The first figure (Churn Class Counts and Proportion) confirms class imbalance in the training data: 3,104 non-churn customers (73.5%) versus 1,121 churn customers (26.5%). This means a model can appear accurate by favoring the majority class, so Recall, F1, and ROC-AUC are more informative than Accuracy alone for this problem.

The second figure (Churn Rate by Contract and Internet Type) shows strong business drivers of churn. Month on Month contracts have the highest churn rate (45.5%), while 1 Yr (10.7%) and 2 year (2.4%) contracts are much more stable. By internet type, high speed fibre optic customers churn the most (40.1%), followed by Cable (24.7%) and DSL (19.8%), with the lowest churn among customers with no internet service (7.2%).

The third figure (Correlation Heatmap) indicates that several numeric features are moderately correlated, especially billing and usage related variables, suggesting overlapping signal in customer behavior. However, no single numeric feature alone explains churn strongly, which supports using multivariate machine learning models rather than relying on one-variable rules. Together, these three views justify the modeling strategy and the focus on targeted retention for high-risk segments.

## Methods of Analysis: Predictive Models
The project compares six supervised models plus a dummy baseline: Dummy Classifier, Logistic Regression, Logistic Regression (Balanced), Random Forest (tuned via GridSearchCV), Extra Trees, Gradient Boosting, and AdaBoost. Model quality is evaluated using Accuracy, Precision, Recall, F1, and ROC-AUC. Because churn is imbalanced, Recall, F1, and ROC-AUC are prioritized over Accuracy. This comparison supports evidence-based selection of the final model.

```python
def evaluate_model(y_true, y_pred, y_prob):
    return {
        "Accuracy": round(accuracy_score(y_true, y_pred), 4),
        "Precision": round(precision_score(y_true, y_pred, zero_division=0), 4),
        "Recall": round(recall_score(y_true, y_pred, zero_division=0), 4),
        "F1": round(f1_score(y_true, y_pred, zero_division=0), 4),
        "ROC_AUC": round(roc_auc_score(y_true, y_prob), 4)
    }
```

```python
# 1) Dummy
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)
dummy_pred = dummy.predict(X_test)
dummy_prob = dummy.predict_proba(X_test)[:, 1]

# 2) Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
pred_lr = lr.predict(X_test)
prob_lr = lr.predict_proba(X_test)[:, 1]

# 3) Logistic Regression (Balanced)
lr_bal = LogisticRegression(max_iter=1000, class_weight="balanced")
lr_bal.fit(X_train, y_train)
pred_lr_bal = lr_bal.predict(X_test)
prob_lr_bal = lr_bal.predict_proba(X_test)[:, 1]

# 4) Random Forest (Grid Search)
rf_base = RandomForestClassifier(class_weight="balanced", random_state=42)
rf_param_grid = {
    "n_estimators": [200, 400],
    "max_depth": [None, 8, 12],
    "min_samples_split": [2, 10],
    "min_samples_leaf": [1, 5]
}
rf_grid = GridSearchCV(
    estimator=rf_base,
    param_grid=rf_param_grid,
    scoring="roc_auc",
    cv=5,
    n_jobs=-1
)
rf_grid.fit(X_train, y_train)
best_rf = rf_grid.best_estimator_
pred_rf = best_rf.predict(X_test)
prob_rf = best_rf.predict_proba(X_test)[:, 1]

# 5) Extra Trees
et = ExtraTreesClassifier(n_estimators=400, random_state=42, class_weight="balanced")
et.fit(X_train, y_train)
pred_et = et.predict(X_test)
prob_et = et.predict_proba(X_test)[:, 1]

# 6) Gradient Boosting
gb = GradientBoostingClassifier(n_estimators=200, learning_rate=0.05, max_depth=4, random_state=42)
gb.fit(X_train, y_train)
pred_gb = gb.predict(X_test)
prob_gb = gb.predict_proba(X_test)[:, 1]

# 7) AdaBoost
ada = AdaBoostClassifier(n_estimators=200, learning_rate=0.5, random_state=42)
ada.fit(X_train, y_train)
pred_ada = ada.predict(X_test)
prob_ada = ada.predict_proba(X_test)[:, 1]

print("Best RF params:", rf_grid.best_params_)
print("Best RF CV ROC-AUC:", round(rf_grid.best_score_, 4))
```

```python
results_map = {
    "Dummy Classifier": (dummy_pred, dummy_prob),
    "Logistic Regression": (pred_lr, prob_lr),
    "Logistic Regression (Balanced)": (pred_lr_bal, prob_lr_bal),
    "Random Forest (Tuned)": (pred_rf, prob_rf),
    "Extra Trees": (pred_et, prob_et),
    "Gradient Boosting": (pred_gb, prob_gb),
    "AdaBoost": (pred_ada, prob_ada),
}

rows = []
for name, (p, pr) in results_map.items():
    metrics = evaluate_model(y_test, p, pr)
    metrics["Model"] = name
    rows.append(metrics)

results_df = pd.DataFrame(rows)[["Model", "Accuracy", "Precision", "Recall", "F1", "ROC_AUC"]]
results_df = results_df.sort_values("ROC_AUC", ascending=False).reset_index(drop=True)
results_df
```

## Results and Findings Explained

The model comparison shows that **Gradient Boosting** is the strongest overall performer, with the highest ROC-AUC (0.9173), highest Accuracy (0.8637), and highest F1-score (0.7257). This indicates strong class discrimination and the best balance between precision and recall among all tested models.

**AdaBoost** is a close second (ROC-AUC 0.9088, F1 0.7151), while **Random Forest (Tuned)** and **Extra Trees** perform reasonably well but with lower recall and/or F1. The baseline **Dummy Classifier** confirms that accuracy alone is misleading in imbalanced churn data, since it produces zero recall and zero F1 for churn cases.

From a business perspective, correctly identifying churners is critical. While **Logistic Regression (Balanced)** achieves the highest recall (0.8182), its precision is much lower (0.4935), which may create too many false alarms. Therefore, for this project, **Gradient Boosting is selected as the final model** because it provides the strongest overall trade-off across ROC-AUC, F1, and practical predictive stability.

```python
plt.figure(figsize=(9, 6))

roc_data = {
    "Logistic Regression": prob_lr,
    "Logistic Regression (Balanced)": prob_lr_bal,
    "Random Forest (Tuned)": prob_rf,
    "Extra Trees": prob_et,
    "Gradient Boosting": prob_gb,
    "AdaBoost": prob_ada
}

for name, pr in roc_data.items():
    fpr, tpr, _ = roc_curve(y_test, pr)
    plt.plot(fpr, tpr, label=f"{name} (AUC={auc(fpr, tpr):.3f})")

plt.plot([0, 1], [0, 1], "k--", label="Chance")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves: Model Comparison")
plt.legend(loc="lower right", fontsize=8)
plt.tight_layout()
plt.show()
```

## Results and Findings Explained
The model comparison table identifies the strongest classifier based on ROC-AUC and F1 performance under class imbalance. Dummy Classifier establishes a baseline and confirms that accuracy alone is insufficient in this problem setting. Tree-based ensembles generally provide stronger discrimination than linear baselines. For business use, the recommended model should prioritize recall and balanced F1 while maintaining acceptable precision.

```python
# Optional threshold tuning for best probability model (business-aligned recall/precision trade-off)
best_model_name = results_df.iloc[0]["Model"]
print("Top model by ROC_AUC:", best_model_name)

best_prob_lookup = {
    "Dummy Classifier": dummy_prob,
    "Logistic Regression": prob_lr,
    "Logistic Regression (Balanced)": prob_lr_bal,
    "Random Forest (Tuned)": prob_rf,
    "Extra Trees": prob_et,
    "Gradient Boosting": prob_gb,
    "AdaBoost": prob_ada
}
best_prob = best_prob_lookup[best_model_name]

thresholds = np.arange(0.20, 0.81, 0.05)
threshold_rows = []
for t in thresholds:
    pred_t = (best_prob >= t).astype(int)
    threshold_rows.append({
        "Threshold": round(float(t), 2),
        "Precision": round(precision_score(y_test, pred_t, zero_division=0), 4),
        "Recall": round(recall_score(y_test, pred_t, zero_division=0), 4),
        "F1": round(f1_score(y_test, pred_t, zero_division=0), 4)
    })

threshold_df = pd.DataFrame(threshold_rows)
threshold_df
```

```python
# Confusion matrix for default threshold on top model
top_pred_map = {
    "Dummy Classifier": dummy_pred,
    "Logistic Regression": pred_lr,
    "Logistic Regression (Balanced)": pred_lr_bal,
    "Random Forest (Tuned)": pred_rf,
    "Extra Trees": pred_et,
    "Gradient Boosting": pred_gb,
    "AdaBoost": pred_ada
}
top_pred = top_pred_map[best_model_name]

cm = confusion_matrix(y_test, top_pred)
print("Confusion Matrix for", best_model_name)
print(cm)
print("\nClassification Report")
print(classification_report(y_test, top_pred, zero_division=0))
```

## Feature and Customer Insights
Feature importance analysis is used to identify operational drivers of churn risk. Variables such as contract type, tenure, and billing-related features are expected to contribute strongly to model decisions. These predictors support targeted interventions rather than broad retention campaigns. In addition, unsupervised clustering is applied to create customer segments for strategy prioritization.

```python
# Feature importance from tuned Random Forest
feat_imp = pd.Series(best_rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
top20 = feat_imp.head(20)

plt.figure(figsize=(10, 7))
top20.sort_values().plot(kind="barh", color="coral")
plt.title("Top 20 Feature Importances (Random Forest)")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

print("Top 10 churn drivers:")
print(top20.head(10).round(4).to_string())
```

```python
# K-means customer segmentation (numeric features only)
numeric_features = churn_train.select_dtypes(include=[np.number]).drop(columns=["Churn"]).columns.tolist()
X_cluster = churn_train[numeric_features].fillna(0)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)

inertia = []
sil_scores = []
K_range = range(2, 9)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertia.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(K_range, inertia, "bo-")
axes[0].set_title("Elbow Method")
axes[0].set_xlabel("k")
axes[0].set_ylabel("Inertia")

axes[1].plot(K_range, sil_scores, "rs-")
axes[1].set_title("Silhouette Score")
axes[1].set_xlabel("k")
axes[1].set_ylabel("Score")
plt.tight_layout()
plt.show()

print("Silhouette scores:")
for k, s in zip(K_range, sil_scores):
    print(f"k={k}: {s:.4f}")
```

```python
# Final cluster profiling (choose k from diagnostics)
best_k = 4
km_final = KMeans(n_clusters=best_k, random_state=42, n_init=10)
churn_train["Cluster"] = km_final.fit_predict(X_scaled)

cluster_summary = churn_train.groupby("Cluster").agg(
    Count=("Churn", "count"),
    Churn_Rate=("Churn", "mean"),
    Avg_Tenure=("Tenure in Months", "mean"),
    Avg_Monthly_Charge=("Monthly Charge", "mean"),
).round(3).reset_index()

cluster_summary
```
