```python
# Import Libraries
import pandas as pd
from sklearn.preprocessing import StandardScaler
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import seaborn as sns
```

**Clustering Analysis on ALS Patient Data**        

In this section, I will apply clustering methods to the ALS patient dataset as outlined in the assignment. 
The goal is to identify meaningful patient groupings that could help in understanding patterns in ALS conditions.

**Import the data**

The first step to generate insights is to understand your data. 
And to understand the data you not just have to understand the variables but also the story behind that data.
I still believe pandas a good in reading data, hence using read_csv function from pandas to read csv file in to a data frame and using head function to read the first 5 rows from it.

```python
# Load the dataset
als_data = pd.read_csv('data//als_data.csv')

# display first few rows to check if data is loaded correctly
als_data.head(5)
```

The ALS dataset used in this analysis contains detailed clinical and laboratory information for each patient, with each row representing a unique individual. The columns include a variety of features such as demographic data (like average age), summary statistics for lab measurements (for example, maximum, minimum, median, and range values for markers like Albumin, Sodium, and Urine pH), and clinical assessments related to ALS progression, such as the ALS Functional Rating Scale  scores and their rate of change. There are also columns capturing specific functional abilities (e.g., trunk strength) summarized in similar ways. Some columns, such as patient IDs, serve only as identifiers and do not provide clinical insight. For clustering and further analysis, the focus will be on the features that directly reflect the patient’s ALS condition and disease progression, while non-informative identifiers will be excluded to ensure meaningful groupings. This rich dataset structure allows for a comprehensive exploration of patterns and subgroups within the ALS patient population.

```python
# Inspect the columns to identify irrelevant features
als_data.info()
als_data.head()

# Drop columns not relevant to ALS condition (e.g., IDs, names, unrelated metadata)
# Adjust the list below based on actual column names in als_data
irrelevant_cols = ['PatientID', 'Name', 'ID', 'SubjectID']  # Example placeholders
als_data_clean = als_data.drop(columns=[col for col in irrelevant_cols if col in als_data.columns])

# Display cleaned data
als_data_clean.head()
```

Lets do some basic EDA on data before moving forward with data standardization.
first check if there are missing values and provide a summary statistics.

```python
als_data_clean.describe()
```

There are no missing values as all features have 2223 rows. There are no outliers when you see max vs mean and 75th percentile values. This means the dataset is now well-prepared for scaling and clustering. All remaining columns represent meaningful patient attributes, such as age, various lab values (e.g., Albumin, Sodium), ALSFRS scores, and functional measures (e.g., trunk strength). These features will allow the clustering algorithm to group patients based on their clinical profiles and disease progression, as intended by the assignment.

As discussed in Chapter 5, it is important to standardize the features so that each contributes equally to the distance calculations. I will use StandardScaler from scikit-learn to scale the data. As Clustering algorithms like K-means are sensitive to the scale of the data, its a good idea to standardize the features.

```python
# Apply standard scalar
scaler = StandardScaler()
als_scaled = scaler.fit_transform(als_data_clean)
```

After applying standard scaling with StandardScaler, the output is a NumPy array, which does not retain the original column names or index from your DataFrame. Lets convert it back to a Data frame to keep the original column names for easier interpretation.

```python
# Convert back to DataFrame for easier handling
als_scaled_df = pd.DataFrame(als_scaled, columns=als_data_clean.columns)
als_scaled_df.head()
```

Now as we have the data frame back, lets create a plot of the cluster silhouette score versus the number of clusters in a K-means cluster. 
Silhouette score measures how well each data point fits within its cluster. Its used to find the best number of clusters. As recommended in the readings, I will plot the silhouette score for a range of cluster counts to visually select the optimal value.

```python
# Test different numbers of clusters and record silhouette scores
silhouette_scores = []
cluster_range = range(2, 11)
for k in cluster_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(als_scaled_df)
    score = silhouette_score(als_scaled_df, labels)
    silhouette_scores.append(score)
optimal_k = cluster_range[np.argmax(silhouette_scores)]
 
 
# Plot silhouette scores
plt.figure(figsize=(8, 5))
plt.plot(cluster_range, silhouette_scores, marker='o')
plt.title('Silhouette Score vs Number of Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```

The plot suggests that the ALS patient data naturally separates into 2 main clusters, with additional clusters not providing better-defined groupings. This helps you justify your choice of the number of clusters for the next steps in your analysis. The score decreases as the number of clusters increases, with a small local peak at k=5.For k > 2, the scores are lower and fluctuate, suggesting that additional clusters do not improve the separation and may even make the clustering less meaningful. So 2 is the optimal number with best score.

With the optimal number of clusters determined as 2, I will fit the K-means model to the scaled data and assign cluster labels to each patient.

```python
# Fit K-means with the optimal number of clusters
kmeans_final = KMeans(n_clusters=optimal_k, random_state=42)
cluster_labels = kmeans_final.fit_predict(als_scaled_df)

# Add cluster labels to the DataFrame
als_scaled_df['Cluster'] = cluster_labels
als_scaled_df.head()
```

To visualize the clusters, I will use Principal Component Analysis (PCA) to reduce the data to two dimensions. This technique, as described in the readings, helps in understanding the structure of high dimensional data.

The PCA function reduce the high dimentional scaled ALS data to two principal components PC1 & PC2. Two components are chosen so the data can be visualized in a 2D scatter plot. fit_transform fits PCA to the scaled data and transform it. It projects each patient in to a new 2D space that captures the most variance in the data for that component.

In the second part we just convert the array back to the data frame

```python
# Fit PCA to reduce to 2 components
pca = PCA(n_components=2, random_state=42)
als_pca = pca.fit_transform(als_scaled_df.drop('Cluster', axis=1))

# Create a DataFrame for plotting
als_pca_df = pd.DataFrame(als_pca, columns=['PC1', 'PC2'])
als_pca_df['Cluster'] = cluster_labels
als_pca_df.head()
```

Now I will create a scatterplot of the PCA-transformed data, coloring each point by its cluster assignment. This visualization helps to see how well the clusters are separated in two dimensions.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=als_pca_df, x='PC1', y='PC2', hue='Cluster', palette='Set2', s=60)
plt.title('PCA Scatterplot of ALS Patients by Cluster')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')
plt.tight_layout()
plt.show()
```

The PCA scatterplot above visualizes the ALS patient data after reducing its high dimensionality to two principal components (PC1 and PC2). Each point represents a patient, and the color indicates the cluster assignment from K-means clustering  with two clusters, as determined optimal by the silhouette score. The plot shows a clear separation between the two clusters along the first principal component, suggesting that the clustering algorithm was able to distinguish two main groups within the patient population. While there is a little overlap, the clusters are generally well-separated, indicating that the features used in the analysis capture meaningful differences among patients.
